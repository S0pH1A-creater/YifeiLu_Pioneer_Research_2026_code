# 9. Risk Management: VaR and Expected Shortfall

- **Objective**: Compute Parametric vs Historical VaR, and introduce Conditional Value at Risk (CVaR).

```python
import numpy as np
import matplotlib.pyplot as plt
import yfinance as yf

S0 = 100000  # $100k Portfolio
mu = 0.08
sigma = 0.2
T = 1.0  
N = 50000

Z = np.random.standard_normal(N)
ST = S0 * np.exp((mu - 0.5 * sigma**2) * T + sigma * np.sqrt(T) * Z)

percentile_5 = np.percentile(ST, 5)
VaR_95 = S0 - percentile_5

print(f'95% Parametric VaR: ${VaR_95:,.2f}')
```

## Conditional Value at Risk (CVaR / Expected Shortfall)

- VaR answers: "What is the minimum I will lose on my worst 5% days?"
- CVaR answers: "When I do breach the VaR threshold, what is my expected average loss?"

```python
tail_events = ST[ST < percentile_5]
CVaR_95 = S0 - np.mean(tail_events)

print(f'95% Conditional VaR (Expected Shortfall): ${CVaR_95:,.2f}')

plt.figure(figsize=(10,5))
counts, bins, patches = plt.hist(ST, bins=80, color='lightgray', edgecolor='black')
plt.axvline(percentile_5, color='orange', linestyle='dashed', linewidth=2, label=f'VaR (95%): ${VaR_95:,.0f} loss')
plt.axvline(np.mean(tail_events), color='red', linestyle='solid', linewidth=2, label=f'CVaR (95%): ${CVaR_95:,.0f} loss')
plt.title('Parametric VaR vs CVaR')
plt.legend()
plt.show()
```

## Historical VaR

- Instead of assuming a mathematical distribution, we can calculate VaR using actual past empirical data.

```python
spy = yf.Ticker('SPY').history(start='2010-01-01', end='2024-01-01')['Close']
daily_returns = spy.pct_change().dropna()

# Sort the actual historical daily returns
historical_5th_percentile = np.percentile(daily_returns, 5)
historical_var_1day = S0 * abs(historical_5th_percentile)

print(f'Empirical 95% 1-Day VaR for a $100k SPY Portfolio: ${historical_var_1day:,.2f}')
```
