# V3 empirical study — 7-day hourly calibration

**This notebook is the computational source of truth.** The PDF in this folder is written from the same `payload` dict.

| Item | Setting |
|------|---------|
| Models | GBM, GARCH, Heston, Merton, GARCH–Merton, Heston–Merton |
| Companies | SPY, AAPL, MSFT |
| Windows | 12 third-Friday monthly expiries, 21 Oct 2022 – 15 Sep 2023 |
| Evaluation | **five NYSE sessions ending the Friday before expiry (hourly 09:59–15:59)** |
| Calibration | **7-day** lookback, **hourly** rolling, 1-minute RTH data |
| LSM horizon | remaining time **to window Friday 15:59**, not listed expiry |
| LSM grid | Δt = **5 minute(s)**, n_steps = remaining RTH / Δt |
| Paths | `10000`, seed `42` |
| Sample | one nearest-ATM listed call per hourly timestamp, DTE 7–60 |

Green / Mark 1 = lowest RMSE%. Friday 15:59 is scheduled but LSM is skipped (`remaining_horizon = 0`).


## Methodology

- Shared sample: nearest-ATM, DTE 7–60, listed expiry, as-of each hourly *t*. All six models price the same contracts.
- No look-ahead: parameters at *t* use only equity returns and listed quotes with timestamp ≤ *t*.
- V3 Heston: option-implied NLS; Euler uses *v_t* then updates variance.
- LSM on the full path cloud. Start every path at the observed *S_t*.

## Findings

- On mean percentage RMSE across the 36 company×window cells, Heston–Merton is the best overall model (mean RMSE% = 84.23; 18 of 36 cells).

- Second overall is Heston (mean RMSE% = 84.27; 15 cells).

- The ranking is not stable. Best-in-cell models across the 36 tables: GBM, Heston, Heston–Merton.

- SPY: lowest mean RMSE% is Heston–Merton (93.17); wins 6 of 12 windows.

- AAPL: lowest mean RMSE% is Heston–Merton (82.34); wins 7 of 12 windows.

- MSFT: lowest mean RMSE% is Heston–Merton (77.17); wins 5 of 12 windows.

- Jump models as a group have lower mean RMSE% (88.76) than the three no-jump models (88.82).

- These conclusions are conditional on the V3 filters, 7-day hourly calibration, nearest-ATM hourly sample, LSM with 10000 paths and 5-minute steps to Friday 15:59, and American-call quotes only. They are not a statement about European options or other tenors.
