# V4 return analysis — 1.5-year monthly study

**This notebook is the computational source of truth.** The matching PDF is written from the same payload.

Realized equity returns in the four windows of `V4 1.5-year fixed empirical study 10000 paths`. This report does not price options and does not resimulate LSM paths.

| Item | Setting |
|------|---------|
| Names | SPY, AAPL, MSFT, AMZN |
| Windows | Crisis, Normal, Late-cycle, COVID (same bounds as the option study) |
| Return | $R_t = \ln(S_t / S_{t-1})$ from `research/data/equity/prices_clean.csv` |
| Table | Mean, Std dev., Skewness statistic, Ex. kurt. statistic, Normality statistics (JB) |
| Brackets | t-test p-values vs Normal for skewness and excess kurtosis; t-version Jarque–Bera p-value |
| Figures | Annualized continuously compounded returns $252 \times R_t$, in percent |


## 2008-2009  ·  Crisis

2008-08-01 → 2009-07-31  ·  n = 252 daily observations.


### Table R1. Return statistics


````python
from IPython.display import display, HTML
display(HTML(ra._html_table(payload['periods'][0])))
````

Notes: This table shows sample statistics for the continuously compounded returns, R_t = ln(S_t / S_{t-1}), for SPY, AAPL, MSFT, and AMZN. The sample period is August 1, 2008 to July 31, 2009, for a total of 252 daily observations. Mean return and standard deviation are in percent per trading day. For the skewness and excess kurtosis statistics, the brackets next to the statistics report the p-values from testing the significance of the difference between the empirical values and the theoretical values from the Normal distribution using a t-test. For the normality statistic the p-value of a t-version of the well known Jarque–Bera test for normality is reported in brackets next to the statistics.


### Figure R1. Annualized continuously compounded returns


````python
fig = ra._draw_return_figure(payload['periods'][0], 1, for_pdf=False)
fig
````

Notes: This figure plots the annualized continuously compounded returns 252 × R_t for SPY, AAPL, MSFT, and AMZN. Date is on the x-axis and the annualized return, in percent, is on the y-axis. Each panel is one name. The shaded band is the evaluation window August 1, 2008 to July 31, 2009 (252 trading days); six months of returns are shown on each side so that band is visible. Large spikes are one-day moves scaled by 252, not cumulative performance.
