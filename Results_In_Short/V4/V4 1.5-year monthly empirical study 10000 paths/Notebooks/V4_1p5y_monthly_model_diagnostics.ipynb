Specification tests on realized returns in the four windows of `V4 1.5-year monthly empirical study 10000 paths`. This report does **not** price options and does **not** use LSM RMSE or MAE.

## Table D1. 2008-2009  ·  Crisis

2008-08-01 → 2009-07-31  ·  n = 252 daily observations.


````python
from IPython.display import display, HTML
display(HTML(dx._html_table(payload['periods'][0])))
````

Notes: This table reports residual diagnostics for GBM, MD-GBM, GARCH, Heston, Merton, GARCH–Merton, Heston–Merton fitted to daily continuously compounded returns R_t = ln(S_t / S_{t-1}) of SPY, AAPL, MSFT, AMZN. The sample is August 1, 2008 to July 31, 2009 (252 trading days). Each model is estimated from that regime and company only. Jump models use standardized residuals after the 3σ jump filter. JB is the usual Jarque–Bera normality test for the standardized residuals. Q(20) is the Ljung–Box portmanteau test for serial correlation in the standardized residuals up to 20 lags; Q²(20) is the same test on the squared standardized residuals. ARCH-LM is Engle's (1982) test for remaining ARCH effects with 5 lags. The test statistic is reported with the p-value in brackets. N/A means the diagnostic is not applicable or the residual series is not well-defined. These tests check model adequacy on returns; they are not option pricing RMSE or MAE.
