# Optimal stopping comparison — rolling = `monthly`

**What this notebook is:** a paste-together of existing study outputs for **one** rolling mode.  
Nothing was re-calibrated or re-priced here.

| Fixed across all 16 blocks | Value |
|----------------------------|-------|
| Lookback | **6 months** |
| Rolling calibration | **`monthly`** |
| Pricing | LSM American calls on SPY (24 contracts, n_paths=2000, seed=42) |

## How to read each block

Each of the **16 studies** has:

| Piece | Meaning |
|-------|---------|
| Heading | **Model · Regime** (and study index 1–16) |
| Metrics table | **RMSE** (primary), MAE, bias, early-exercise — for **this** model, **this** regime, rolling=`monthly` |
| Three-panel figure | §6 LSM graph for that same group: **left** model vs market · **middle** mean option value · **right** exercise timing |

**Order:** regimes `2008-2009` → `2013-2014` → `2018-2019` → `2019-2020`; within each regime: GBM → Merton → Heston–Merton → GARCH–Merton.

Sibling notebooks (same 16 studies, different rolling only):

- `compare_rolling_none.ipynb`
- `compare_rolling_monthly.ipynb`
- `compare_rolling_daily.ipynb`
