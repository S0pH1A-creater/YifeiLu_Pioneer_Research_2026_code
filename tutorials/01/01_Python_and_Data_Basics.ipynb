# 1. Python and Data Basics

- **Objective**: Extract, clean, and visualize multi-dimensional financial data.

## Multi-Asset Data Extraction

- Extract historical data for multiple assets: Apple (AAPL), Microsoft (MSFT), and the S&P 500 (SPY).

```python
import yfinance as yf # stock data
import pandas as pd # data analysis tools
import matplotlib.pyplot as plt # graphing tools

tickers = ['AAPL', 'MSFT', 'SPY']
# Download data and keep only the 'Close' prices
data = yf.download(tickers, start='2023-01-01', end='2024-01-01')['Close']
print(data.head())
```

## Data Cleaning

- Financial data often contains missing values (NaN) due to market holidays or data errors.
- We use forward-fill (`ffill()`) to propagate the last valid observation forward, preventing gaps.

```python
data = data.ffill().dropna()
print(f'Missing values after cleaning: {data.isna().sum().sum()}')
```

## Normalization and Visualization

- Comparing stocks with drastically different prices requires normalization (setting the starting price to 1.0).

```python
normalized_data = data / data.iloc[0] # this is the stock price divided by the INITIAL stock price, S0
plt.figure(figsize=(12,6))
for col in normalized_data.columns:
    plt.plot(normalized_data[col], label=col)

plt.title('Normalized Price Performance (2023)')
plt.ylabel('Growth (1.0 = Initial Value)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

## Rolling Transformations

- Compute a 20-day Simple Moving Average (SMA) for AAPL to smooth out noise.

```python
aapl = data['AAPL']
sma_20 = aapl.rolling(window=20).mean()

plt.figure(figsize=(12,6))
plt.plot(aapl, label='AAPL Price', alpha=0.5)
plt.plot(sma_20, label='20-Day SMA', color='red', linewidth=2)
plt.title('AAPL Price vs 20-Day Simple Moving Average')
plt.legend()
plt.show()
```
