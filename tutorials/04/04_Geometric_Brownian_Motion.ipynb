# 4. Geometric Brownian Motion (GBM)

- **Objective**: Mathematically decompose the GBM equation and compare simulated paths with empirical data.

## The GBM Formula

$$S_T = S_0 \exp\left(\left(\mu - \frac{\sigma^2}{2}\right)T + \sigma \sqrt{T} Z\right)$$

- $S_0$: Initial price
- $\mu$: Expected return (drift)
- $\sigma$: Volatility
- $Z$: Random normal variable

## Decomposing Drift and Diffusion

- Let's separate the deterministic drift component from the stochastic diffusion (noise) component to see how they combine.

```python
import numpy as np
import matplotlib.pyplot as plt

S0 = 100
mu = 0.08
sigma = 0.2
T = 1
N_steps = 252
dt = T / N_steps

# Arrays for time and components
times = np.linspace(0, T, N_steps)
Z = np.random.standard_normal(N_steps)
W = np.cumsum(Z) * np.sqrt(dt)  # Brownian Motion path

# 1. Pure Drift Component (Deterministic)
drift = (mu - 0.5 * sigma**2) * times

# 2. Pure Diffusion Component (Stochastic)
diffusion = sigma * W

# 3. Combined GBM Path
path = S0 * np.exp(drift + diffusion)

fig, (ax1, ax2, ax3) = plt.subplots(3, 1, figsize=(10, 10))
ax1.plot(times, drift, color='green')
ax1.set_title('1. Deterministic Drift: $(\mu - \sigma^2/2)T$')

ax2.plot(times, diffusion, color='purple')
ax2.set_title('2. Stochastic Diffusion: $\sigma \sqrt{T} Z$')

ax3.plot(times, path, color='darkorange')
ax3.set_title('3. Combined Exponential Path: $S_T$')

plt.tight_layout()
plt.show()
```

## Empirical Comparison

- How realistic is this? Let's plot a GBM simulation alongside an actual historical path of a stock (AAPL in 2023).

```python
import yfinance as yf
import pandas as pd

aapl = yf.Ticker('AAPL').history(start='2023-01-01', end='2024-01-01')['Close']
# Normalize AAPL to start at 100
aapl_norm = (aapl / aapl.iloc[0]) * 100

plt.figure(figsize=(12,6))
plt.plot(aapl_norm.values, label='Actual AAPL (2023)', color='black', linewidth=2)
plt.plot(path, label='Simulated GBM Path', color='orange', alpha=0.8)
plt.title('Reality vs. Simulation')
plt.legend()
plt.show()
```
