## Basic Call and Put Options

- **Call Option**: The right to BUY the stock at Strike Price $K$.
- **Put Option**: The right to SELL the stock at Strike Price $K$.
- **Premium ($C$ or $P$)**: The upfront cost to buy the option.
- **Profit**: Payoff - Premium

```python
import numpy as np
import matplotlib.pyplot as plt

ST = np.linspace(50, 150, 200)
K = 100       # Strike price
premium = 5   # Upfront cost of the option

call_profit = np.maximum(ST - K, 0) - premium
put_profit = np.maximum(K - ST, 0) - premium

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

ax1.plot(ST, call_profit, color='green', lw=2)
ax1.axhline(0, color='black', linestyle='--')
ax1.axvline(K + premium, color='blue', linestyle=':', label=f'Break-Even (${K + premium})')
ax1.set_title('Call Option Profit at Expiration')
ax1.legend()

ax2.plot(ST, put_profit, color='red', lw=2)
ax2.axhline(0, color='black', linestyle='--')
ax2.axvline(K - premium, color='blue', linestyle=':', label=f'Break-Even (${K - premium})')
ax2.set_title('Put Option Profit at Expiration')
ax2.legend()

plt.show()
```

## Option Moneyness

- **ITM (In The Money)**: Option has intrinsic value.
- **ATM (At The Money)**: Strike price roughly equals current stock price.
- **OTM (Out of The Money)**: Option has zero intrinsic value.

```python
plt.figure(figsize=(10,5))
plt.plot(ST, call_profit, color='black', lw=2)
plt.fill_between(ST, call_profit, -premium, where=(ST > K), color='green', alpha=0.3, label='ITM (In The Money)')
plt.fill_between(ST, call_profit, -premium, where=(ST <= K), color='red', alpha=0.3, label='OTM (Out of The Money)')
plt.axhline(0, color='black', linestyle='--')
plt.axvline(K, color='purple', linestyle='--', label=f'ATM (Strike $K={K})')
plt.title('Call Option Moneyness')
plt.legend()
plt.show()
```

## Multi-Asset Strategy: The Protective Put

- **Strategy**: Buy the underlying stock ($S_0$) + Buy a Put Option to protect against a crash.
- This effectively truncates your downside risk while preserving unlimited upside.

```python
S0 = 100
stock_profit = ST - S0
put_premium = 6
protective_put_profit = stock_profit + np.maximum(K - ST, 0) - put_premium

plt.figure(figsize=(10,6))
plt.plot(ST, stock_profit, label='Long Stock Only', color='black', linestyle='--')
plt.plot(ST, protective_put_profit, label='Protective Put (Stock + Put)', color='blue', linewidth=3)
plt.axhline(0, color='gray', linestyle='-')
plt.axvline(S0, color='gray', linestyle=':')
plt.title('Risk Transformation: The Protective Put')
plt.xlabel('Stock Price at Expiration')
plt.ylabel('Profit')
plt.legend()
plt.show()
```
